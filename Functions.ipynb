```python
#Iterators and Generators
#Iterators and generators are both used to process data one item at a time instead of loading everything into memory at once.
```

```python
#ITERATORS: An iterator is an object that gives one value at a time.
#Instead of getting all values together, you ask for the next value whenever you need it.
#Python uses the next() function to get the next item.
num=[1,2,3,4]
x=iter(num)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

```python
#GENERATOR:A generator is a special kind of function that creates an iterator automatically.
#Generator uses yield instead of return
#because return ends the function immediately and returns the value immediately
#where as yield sends one value, pauses the function, remembers where it stopped and continues from there the next time next() is called
```

```python
def numbers(num1,num2):
    return num1 #after this value it stops and only returns num1
    return num2
print(numbers(2,3))
```

```python
def numbers(num1,num2,num3):
    yield num1 #function doesn't restart every time it remembers where it stopped
    yield num2
    yield num3

g = numbers(1,2,3)

print(next(g))
print(next(g))
print(next(g))
```

```python
def numbers():
    for i in range(1000):
        yield i
x=numbers()
print(next(x))
print(next(x))
print(next(x))
```

```python
def even_numbers():
    for num in range(2, 11, 2):
        yield num

g = even_numbers()

for num in g:
    print(num)
```

```python
def even_numbers():
    for num in range(2, 11, 2):
        yield num

g = even_numbers()

print(next(g))
print(next(g))
print(next(g))

for num in g:
    print(num)
```

```python
def fibonacci(limit):

    a = 0
    b = 1

    for i in range(limit):
        yield a
        a, b = b, a + b


fib = fibonacci(10)
print(next(fib))
print(next(fib))
for num in fib:
    print(num)
```

```python
#Lambda Function:A lambda function is a small anonymous function (a function without a name)

square = lambda x: x * x
print(square(5))
add = lambda a,b: a+b
print(add(2,3))
```

```python
#map(): applies a function to every item in a collection.
numbers = [1, 2, 3, 4]

result = []

for num in numbers:
    result.append(num * num)

print(result)
```

```python
numbers=[1,2,3,4]
result=map(lambda x:x*x,numbers)
print(result)
print(list(result))
```

```python
#filter():keeps only the elements that satisfy a condition(returns only true values)
num=[1,2,3,4,5,6,7,8]
result=filter(lambda x: x%2==0,num)
print(result)
print(list(result))
```

```python
#reduce():combines all the elements into one final value.

numbers=[1,2,3,4]
result=reduce(lambda a,b:a+b,numbers)
print(result)
```

```python
num=[2,9,0,23,100]
largest=reduce(lambda a,b:a if a>b else b,num)
print(largest)
```

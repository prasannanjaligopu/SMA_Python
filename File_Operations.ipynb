```python
with open(r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'w') as f:
    f.write("Hello this is Anjali\n")
    f.write("I'writing some sample text file")

with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r') as x:
    content=x.read()
    print(content)
```

```python
#'a' represents append, we use append because if we use write function, it overwrites the contents that are already
#existed in that file and write the contents that we gave
#whereas append will just adds the content
with open(r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'a') as f:
    f.write("\nThis is just for practice")

with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r') as x:
    content=x.read()
    print(content)
```

```python
#Each call to readline() reads the next line.
with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r') as x:
    print(x.readline())
    print(x.readline())
```

```python
#readlines: It reads all lines as a list
with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r') as x:
    print(x.readlines())
```

```python
#r+: It will helps to read and write the files at a time
with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r+') as x:
    content=x.read()
    print(content)
    x.write("\nLet's make this topic little bit intresting")
    
```

```python
#w+: It will helps to write and read the file at a time
with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'w+') as x:
    x.write("\nLet's add some more date")
    x.seek(0)
    content=x.read()
    print(content)
```

```python
#a+: will helps to add the data and read the data at a time
with open (r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'a+') as x:
    x.write("\nLearn File handling")
    x.seek(0)
    content=x.read()
    print(content)
```

```python
with open(r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt", "r+") as x:

    print("Before adding data:")
    content = x.read()
    print(content)

    x.write("\nThis is the added line")

    x.seek(0)

    print("\nAfter adding data:")
    print(x.read())     
```

```python
#Copying the content of the files:
with open(r"C:\Users\prasa\OneDrive\Desktop\Class\Sample.txt",'r') as s, open(r"C:\Users\prasa\OneDrive\Desktop\Class\Example.txt",'w') as d:
    for line in s:
        d.write(line)

with open(r"C:\Users\prasa\OneDrive\Desktop\Class\Example.txt",'r') as x:
    content=x.read()
    print(content)
```

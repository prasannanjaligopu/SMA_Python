```python
#Creating Strings
name= "Prasannanjali"
print(name)
Initial='Gopu'
print(Initial)

x="""This is a 
multi line string
"""
print(x)

#Empty String
empty=""
```

```python
text='Python'

print(text[0])
print(text[5])
#negative indexex
print(text[-1])
print(text[-3])

#slicing
print(text[:])
print(text[0:])
print(text[:6])
print(text[1:4])
print(text[::2])
print(text[::-1]) #reversing the string
```

```python
#Concatination
str1='Hello'
str2='World'
con1=str1+" "+str2
con2=str1+"."+str2
print(con1)
print(con2)
text="Hello"
print(text+" "+"World")
print(text) #in previos line we concatinated but that is only for that line, it won't change the actual string
text=text+" World"  #when we assign the string using assignment operator then it will update the contents of the string
print(text)
```

```python
#Repetetion
x="Ha"*3
print(x)
```

```python
#Using f-string
age=25
name='Anjali'
#normally we can not concatinate string with integers, to concatenate these we use f-strings
message=f"{name} is {age} years old"
print(message)
# Using format()
message = "{} is {} years old".format(name, age)
print(message)
# Using % formatting
message=f"%s is %d years old"%(name,age)
print(message)
```

```python
text="Hello World Python"
#case conersion
print(text.lower())
print(text.upper())
print(text.capitalize())
print(text.title())
print(text.swapcase())
#replace
print(text.replace("World","Python"))
print(text)

#Split and Join
x=text.split()
print(x)
joined="-".join(x)
print(joined)
print(text)
```

```python
text = "Hello World Python"
print(text.find("World"))
print(text.find("xyz"))
print(text.index("Python")) #same like find but returns error if the string not found
print(text.count("o"))
# Checking content
print(text.startswith("Hello"))  
print(text.endswith("Python"))   
print(text.isdigit())            
print("123".isdigit())          
print("abc".isalpha())           
print("abc123".isalnum())  
```

```python
# Whitespace operations
messy = "  Hello World  "
print(messy)
print(messy.strip()) #strips both sides
print(messy.lstrip()) #left only
print(messy.rstrip()) #right only
```

```python
print("Anjali".center(12))
print("Anjali".ljust(12))
print("Anjali".rjust(12))
```

```python
#LIST:A list is an ordered, mutable collection that can store multiple items of any type, including duplicates.
num=[1,3.5,5,6,6,'xyz',True,None]
print(num)

# Nested lists (lists within lists)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix)
```

```python
#Accessing Elements
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[0])
print(fruits[2])
print(fruits[-1])
print(fruits[:])
print(fruits[0:])
print(fruits[:4])
print(fruits[0:2])
print(fruits[::-1]) #reversing the list
print(fruits[::2])
```

```python
#ADDING Elements to the list
fruits = ["apple", "banana"]
#Adding to the list
fruits.append('cherry')
print(fruits)
#Adding element at particular index
fruits.insert(1,'Strawberry')
print(fruits)
#Adding multiple elements
fruits.extend(['apple','watermelon'])
print(fruits)

fig=['dates','oranges']
fruits.insert(1,fig)
print(fruits)
fruits.append(fig)
print(fruits)

fruits+=fig
print(fruits)
```

```python
#CHANGING the elements
colors = ["red", "green", "blue"]
colors[1]='orange'
print(colors)

colors[1:3] = ["purple", "orange"]
print(colors)
```

```python
#REMOVE Elements
items = ["a", "b", "c", "d", "e","f"]
# remove() - Remove by value
items.remove('c')
print(items)
# pop() - Remove by index
removed = items.pop()
print(removed)
print(items)

removed = items.pop(0) 
print(removed)
print(items)

# del - Delete by index
del items[0]
print(items)

# clear() - Remove all elements
items.clear()
print(items)
```

```python
#SORTING:
numbers = [3, 1, 4, 1, 5, 9, 2]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)

original=['b','c','a']
sorted_of_original=sorted(original)
print(original)
print(sorted_of_original)
```

```python
#SHALLOW COPY
original=[1,2,3]
copy1=original.copy()

copy1[0]=99
print(original)  #original value unchanged
print(copy1)

ref=original
ref[0]=99
print(ref)
print(original)  #original value was changed
```

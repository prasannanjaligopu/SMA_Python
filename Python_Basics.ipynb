```python
SQL: A declarative language specifically for querying, manipulating, and managing data in databases
Python: A general-purpose programming language for writing applications, automation, data analysis, machine learning, web development, etc.
    - Python is a high-level, general-purpose programming language that emphasizes readability and simplicity.
    #High-level language is one that abstracts away the computer's low-level details. High-level means easy to read and write but slower.
    - You write code that reads like human logic, and the language/interpreter handles the technical complexity behind the scenes.
    #Dynamic language: A dynamic language determines data types at runtime (when code runs), not at compile time (before running). 
     #You don't need to declare what type a variable is — Python figures it out.
    #Object-Oriented Programming Language:Object-Oriented Programming is a way of organizing code around "objects"
    — things that bundle data (attributes) and actions (methods) together.
    #Benefits of OOP:
    Organize code logically — related data and functions together
    Reusable — write once, use multiple times
    Maintainable — easier to update and debug
    Scalable — great for large projects
    Real-world modeling — code mirrors how we think about things
```

```python
#Fundamental Data types:
#1.Integer:
a=9
b=10
a=11
print(a+b) # It stores the latest value 

print(int(3.14))
print(int('10'))
print(int(True))
print(int(False))
```

```python
a=9
print(type(a))
b='Anjali'
print(type(b))
```

```python
# Type casting: It is the process of converting a value from one data type to another.
# Implicit: It automatically converts a lower type to a higher type to avoid data loss. This happens without you explicitly asking.
a=9
b=10.9
print(a+b)
```

```python
#Explicit: explicitly tell Python to convert a type using conversion functions.
a=10
b='Anjali'
#print(a+b) here we cannot add the integer to string, so we convert the data type and adds
print(str(a)+b)
```

```python
#2.Float:Numbers with decimal points. Used for precise, fractional values.
price=90
discounted_price=price*0.11
print(discounted_price)
print(type(discounted_price))

print(float(10))
print(float(True))
```

```python
#3.Strings:Text — anything enclosed in quotes. Single ', double ", or triple """ quotes.
a='John'
print(a)
print(type(a))

full_name="Alice" + " " +a
print(full_name)

text="PYTHON"
print(text.lower())
print(text.replace("P","Z"))
print(text)
```

```python
text='Python'
print(text[0])      
print(text[2])      
print(text[-1])     
print(text[-2])    
print(text[0:3])    
print(text[1:4])    
print(text[:3])     
print(text[2:])     
print(text[::2])    
print(text[::-1]) #Reversing
```

```python
#4.Boolean:Two values only: True or False. Used for conditions and logic.
x=5
y=10
print(x>y)
print(x<y)

print(x==y)
print(x<=y)
print(x!=y)
```

```python
#5.None:It represents the absence of a value. It's a placeholder for "nothing."
x=[1,2,3,4]
print(x(5))
```

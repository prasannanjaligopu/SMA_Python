```python
#1. ATM PIN Verification using for loop
correct_pin = 1234
max_attempts = 3

for attempt_number in range(max_attempts):
    entered_pins = input("Enter PIN: ")
    
    current_pin = int(entered_pins)
    
    if current_pin == correct_pin:
        print("Access Granted")
        break
    tries_left = max_attempts - attempt_number - 1
    
    if tries_left > 0:
        print(f"Wrong PIN, {tries_left} tries left")
    else:
        print("Card Locked")
else:
    print("Card Locked - Maximum attempts exceeded")
```

```python
# 1. ATM PIN Verification using while loop
correct_pin = 1234
max_attempts = 3
attempt_number = 0

while attempt_number < max_attempts:
    entered_pin = int(input("Enter PIN: "))
    
    if entered_pin == correct_pin:
        print("Access Granted!")
        break
    attempt_number += 1
    
    tries_left = max_attempts - attempt_number
    
    if tries_left > 0:
        print(f"Wrong PIN. {tries_left} tries left.")
    else:
        print("Card Locked - Too many failed attempts!")
```

```python
# 2.Grocery Checkout Total using for loop

n = int(input("Enter the number of items: "))

total = 0
discount = False

for i in range(n):
    price = float(input(f"Enter price of item {i + 1}: "))

    if discount:
        price = price * 0.9

    total += price

    if total >= 1000:
        discount = True

print(f"Final Total: ₹{total:.2f}")
```

```python
# 2.Grocery Checkout Total using while loop
n = int(input("Enter number of items: "))

total = 0
i = 0

while i < n:
    price = float(input(f"Enter price of item {i + 1}: "))

    if total + price > 1000:
        price = price * 0.9

    total += price

    print("Running Total =", total)

    i += 1

print("Final Total =", total)
```

```python
# 3.Number Guessing Game using for loop

secret_number = int(input("Enter the secret number: "))
n = int(input("Enter the number of guesses: "))

guesses = []

for i in range(n):
    guess = int(input(f"Enter guess {i + 1}: "))
    guesses.append(guess)

count = 0

for guess in guesses:
    count += 1

    if guess > secret_number:
        print("Too High")
    elif guess < secret_number:
        print("Too Low")
    else:
        print(f"Correct! You guessed it in {count} guesses.")
        break
```

```python
# 3.Number Guessing Game using while loop

secret_number = int(input("Enter the secret number: "))
n = int(input("Enter the number of guesses: "))

guesses = []

i = 0
while i < n:
    guess = int(input(f"Enter guess {i + 1}: "))
    guesses.append(guess)
    i += 1

count = 0
i = 0

while i < len(guesses):
    count += 1

    if guesses[i] > secret_number:
        print("Too High")
    elif guesses[i] < secret_number:
        print("Too Low")
    else:
        print(f"Correct! You guessed it in {count} guesses.")
        break

    i += 1
```

```python
# 4.Movie Ticket Booking using for loop

total_seats = int(input("Enter total number of seats: "))
n = int(input("Enter the number of booking requests: "))

for i in range(n):
    request = int(input(f"Enter seats requested for booking {i + 1}: "))

    if request <= total_seats:
        total_seats -= request
        print(f"Booking successful! Seats remaining: {total_seats}")

        if total_seats == 0:
            print("All seats are sold!")
            break
    else:
        print("Not enough seats")
        break

print("Seats remaining:", total_seats)
```

```python
# 5.Study Streak Tracker using for loop

n = int(input("Enter the number of quiz scores: "))

streak = 0

for i in range(n):
    score = int(input(f"Enter score for day {i + 1}: "))

    if score >= 60:
        streak += 1
    else:
        print("Streak broken")
        break

print("Final streak length:", streak)
```
